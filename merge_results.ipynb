```python


#to merge pretrained.jsonl / sft.jsonl / dpo.jsonl into one


import json
import pandas as pd

RESULTS_DIR = "/content/drive/MyDrive/Colab Notebooks/pragmatics_project/results"
STAGES = ["pretrained", "sft", "dpo"]

rows = []
for stage in STAGES:
    path = f"{RESULTS_DIR}/{stage}.jsonl"
    with open(path) as f:
        for line in f:
            r = json.loads(line)
            rows.append(r)

df = pd.DataFrame(rows)
print(f"Total rows loaded: {len(df)}")
print(f"Rows with errors: {df['error'].notna().sum()}")

# keep only reliable steps, drop any error rows
clean_df = df[(df["reliable"] == True) & (df["error"].isna())].copy()
print(f"Clean, reliable rows: {len(clean_df)} (expect 24 items x 4 conditions x 2 steps x 3 checkpoints = 576)")

#  winner_meaning to a simple binary outcome for analysis
clean_df["outcome_likely"] = (clean_df["winner_meaning"] == "likely").astype(int)

clean_df.to_csv("//content/drive/MyDrive/Colab Notebooks/pragmatics_project/results/merged_clean.csv", index=False)
print("Saved -> /content/drive/MyDrive/Colab Notebooks/pragmatics_project/results/merged_clean.csv")
print(clean_df[["stage", "item_id", "condition", "step", "winner_meaning"]].head(10))
```
